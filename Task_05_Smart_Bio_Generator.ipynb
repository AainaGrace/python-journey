```python
# Task 05 Smart Bio Generator

name = input("Enter your full name: ")
course = input("Enter your course: ")
skills = input("Enter your top skills: ")
projects = input("Enter a project: ")
social_link = input("Enter your LinkedIn or Githhub URL: ")



#LinkedIn Bio
print("=" * 40)
print("      LINKEDIN BIO")
print("=" * 40)
print(f"Name        : {name}")
print(f"Course      : {course}")
print(f"Skills      : {skills}")
print(f"Projects    : {projects}")
print(f"Link        : {social_link}")

#GITHUB BIO
print("=" * 40)
print("      GITHUB BIO")
print("=" * 40)
print(f"User        : {name} | {course}")
print(f"Skills      : {skills}")
print(f"Projects    : {projects}")
print(f"Portfoolio  : {social_link}")

#TWITTER / X BIO
print( "=" * 40)
print("       TWITTER / X BIO")
print( "=" * 40)
print(f"Bio       : {name} | {course}")
print(f"Focus     : Building {projects}")
print(f"Skills    : {skills}")
print(f"Connect   : {social_link}")

# PERSONAL WEBSITE BIO
print( "=" * 40)
print("       PERSONAL WEBSITE BIO")
print( "=" * 40)
print(f"About Me  : Hi! I'm {name}, specializing in {course}.")
print(f"Current   : Focused on {skills} and developing {projects}.")
print(f"Contact   : Connect with me {social_link}")
print("=" * 40)
```
